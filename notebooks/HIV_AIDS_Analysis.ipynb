# PHÂN TÍCH DỮ LIỆU HIV/AIDS TOÀN CẦU VÀ SO SÁNH CÁC MÔ HÌNH HỒI QUY TRONG DỰ BÁO CÁC CHỈ SỐ HIV/AIDS

# CHƯƠNG 1. TỔNG QUAN ĐỀ TÀI

## 1.1. Lý do chọn đề tài

HIV/AIDS là một trong những vấn đề sức khỏe cộng đồng nhận được sự quan tâm trên phạm vi toàn cầu. HIV là virus gây suy giảm miễn dịch ở người, trong khi AIDS là hội chứng suy giảm miễn dịch mắc phải có thể xuất hiện ở giai đoạn tiến triển của tình trạng nhiễm HIV nếu người bệnh không được điều trị hiệu quả.

Mặc dù đã có nhiều tiến bộ trong công tác phòng ngừa, xét nghiệm và điều trị, HIV/AIDS vẫn đặt ra nhiều thách thức đối với hệ thống y tế. Việc theo dõi số người sống chung với HIV, số ca nhiễm mới, số ca tử vong liên quan đến AIDS và tỷ lệ bao phủ điều trị bằng thuốc kháng virus có vai trò quan trọng trong việc tìm hiểu diễn biến của dịch bệnh.

Các chỉ số HIV/AIDS có thể khác nhau giữa các quốc gia, khu vực và thời điểm. Vì vậy, việc thu thập, xử lý và phân tích dữ liệu là cần thiết nhằm nhận diện xu hướng, tìm hiểu đặc điểm của dữ liệu và hỗ trợ đánh giá tình hình dịch bệnh.

Sự phát triển của công nghệ thông tin và học máy tạo điều kiện ứng dụng các phương pháp phân tích dữ liệu vào những vấn đề thực tiễn. Với ngôn ngữ Python và các thư viện như Pandas, NumPy, Matplotlib và Scikit-learn, người nghiên cứu có thể thực hiện quy trình từ tiền xử lý dữ liệu, phân tích khám phá đến xây dựng và đánh giá mô hình dự báo.

Xuất phát từ những lý do trên, đề tài “Phân tích dữ liệu HIV/AIDS toàn cầu và so sánh các mô hình hồi quy trong dự báo các chỉ số HIV/AIDS” được thực hiện nhằm khảo sát ba bộ dữ liệu liên quan đến HIV/AIDS, phân tích các đặc điểm nổi bật và so sánh hiệu quả của năm mô hình hồi quy trên những dữ liệu phù hợp. Qua đó, đề tài vận dụng kiến thức lập trình Python, phân tích dữ liệu và học máy vào một vấn đề có ý nghĩa đối với sức khỏe cộng đồng.

## 1.2. Mục tiêu nghiên cứu

### 1.2.1. Mục tiêu tổng quát

Phân tích ba bộ dữ liệu liên quan đến HIV/AIDS, khám phá đặc điểm và xu hướng của các chỉ số quan trọng, đồng thời xây dựng và so sánh năm mô hình hồi quy nhằm đánh giá khả năng dự báo các chỉ số định lượng phù hợp với dữ liệu.

### 1.2.2. Mục tiêu cụ thể

- Khảo sát cấu trúc, nội dung và chất lượng của ba bộ dữ liệu.
- Tiền xử lý dữ liệu, kiểm tra giá trị thiếu, dữ liệu trùng lặp và những sai lệch về định dạng.
- Thực hiện thống kê mô tả và trực quan hóa các chỉ số HIV/AIDS.
- Lựa chọn biến đầu vào và biến mục tiêu phù hợp với từng bộ dữ liệu.
- Xây dựng năm mô hình hồi quy: Linear Regression, Ridge Regression, Random Forest Regressor, Gradient Boosting Regressor và Support Vector Regression (SVR).
- Đánh giá và so sánh mô hình bằng các chỉ số MAE, RMSE và R².
- Rút ra nhận xét dựa trên kết quả thực nghiệm và đề xuất hướng phát triển.

## 1.3. Câu hỏi nghiên cứu

Đề tài tập trung trả lời các câu hỏi sau:

1. Ba bộ dữ liệu có đặc điểm gì về cấu trúc, phạm vi thời gian và các chỉ số HIV/AIDS?
2. Các chỉ số HIV/AIDS có những đặc điểm phân bố và xu hướng nào trong phạm vi dữ liệu?
3. Những biến nào phù hợp để sử dụng làm biến đầu vào và biến mục tiêu cho các mô hình hồi quy?
4. Năm mô hình hồi quy đạt hiệu quả dự báo như thế nào trên từng bộ dữ liệu phù hợp?
5. Mô hình nào có kết quả tốt hơn khi được đánh giá bằng MAE, RMSE và R² trên cùng một tập dữ liệu kiểm tra?
6. Những hạn chế về chất lượng dữ liệu và phương pháp đánh giá ảnh hưởng như thế nào đến kết quả nghiên cứu?

## 1.4. Đối tượng và phạm vi nghiên cứu

### 1.4.1. Đối tượng nghiên cứu

Đối tượng nghiên cứu là dữ liệu thống kê liên quan đến HIV/AIDS và hiệu quả của các mô hình hồi quy trong dự báo các chỉ số định lượng.

Ba bộ dữ liệu được sử dụng gồm:

- HIV_AIDS prevalence estimates table.csv: chứa các thông tin về tỷ lệ hiện nhiễm, số người sống chung với HIV/AIDS, số ca tử vong hằng năm và năm ước tính.
- aids_regional_aggregates.csv: chứa các thông tin về quốc gia, năm, dân số, một số chỉ số kinh tế - xã hội và các chỉ số HIV/AIDS.
- art_coverage_by_country_clean.csv: tập trung vào số người nhận điều trị ART, số người ước tính sống chung với HIV và tỷ lệ bao phủ điều trị ART.

Các thuộc tính cụ thể sẽ được kiểm tra trực tiếp từ những tệp CSV trong quá trình thực hiện.

### 1.4.2. Phạm vi nội dung

Đề tài tập trung vào việc khảo sát và tiền xử lý dữ liệu, phân tích khám phá dữ liệu, xây dựng năm mô hình hồi quy và đánh giá kết quả dự báo. Nghiên cứu không nhằm thay thế các nghiên cứu dịch tễ học chuyên sâu hoặc sử dụng kết quả mô hình để chẩn đoán và điều trị cho cá nhân.

### 1.4.3. Phạm vi dữ liệu

Phạm vi địa lý hướng đến dữ liệu quốc gia và khu vực trên toàn cầu trong phạm vi mà các bộ dữ liệu cung cấp. Phạm vi thời gian được xác định theo những năm thực tế xuất hiện trong từng bộ dữ liệu. Do các bộ dữ liệu có thể khác nhau về phạm vi và cấu trúc, việc phân tích và xây dựng mô hình sẽ được điều chỉnh cho phù hợp.

## 1.5. Phương pháp nghiên cứu

### 1.5.1. Khảo sát dữ liệu

Tiến hành đọc và kiểm tra ba bộ dữ liệu, xác định số lượng dòng, số lượng cột, tên thuộc tính, kiểu dữ liệu và phạm vi quan sát. Đồng thời, đánh giá mức độ phù hợp của từng bộ dữ liệu đối với mục tiêu phân tích và dự báo.

### 1.5.2. Tiền xử lý dữ liệu

Kiểm tra giá trị thiếu, bản ghi trùng lặp, dữ liệu không hợp lệ và sai lệch định dạng. Thực hiện chuyển đổi kiểu dữ liệu, xử lý giá trị thiếu và mã hóa biến phân loại khi cần thiết. Những thao tác tiền xử lý được lựa chọn dựa trên đặc điểm của từng bộ dữ liệu nhằm hạn chế làm sai lệch thông tin.

### 1.5.3. Phân tích khám phá dữ liệu

Sử dụng thống kê mô tả và biểu đồ để tìm hiểu sự phân bố của các chỉ số HIV/AIDS, sự thay đổi theo thời gian và sự khác biệt giữa các nhóm quan sát khi dữ liệu cho phép. Kết quả được trình bày bằng bảng thống kê và hình ảnh trực quan.

### 1.5.4. Xây dựng mô hình hồi quy

Đề tài dự kiến sử dụng năm mô hình:

1. Linear Regression: mô hình hồi quy tuyến tính cơ sở.
2. Ridge Regression: hồi quy tuyến tính có điều chuẩn L2.
3. Random Forest Regressor: mô hình tổ hợp nhiều cây quyết định.
4. Gradient Boosting Regressor: mô hình tăng cường tuần tự.
5. Support Vector Regression (SVR): mô hình hồi quy dựa trên véc-tơ hỗ trợ.

Việc lựa chọn biến mục tiêu và thiết lập mô hình được thực hiện dựa trên cấu trúc dữ liệu thực tế. Các mô hình được huấn luyện và đánh giá theo quy trình phù hợp nhằm hạn chế rò rỉ dữ liệu.

### 1.5.5. Đánh giá mô hình

Các mô hình được đánh giá bằng ba chỉ số:

- MAE: đo sai số tuyệt đối trung bình; giá trị thấp hơn thường thể hiện dự báo gần giá trị thực tế hơn.
- RMSE: đo căn bậc hai của sai số bình phương trung bình; nhạy hơn với các sai số lớn.
- R²: phản ánh mức độ phù hợp của dự báo so với mức tham chiếu dựa trên giá trị trung bình của biến mục tiêu.

Việc lựa chọn mô hình phù hợp căn cứ vào kết quả thực nghiệm, các chỉ số đánh giá và đặc điểm của từng bộ dữ liệu.

### 1.5.6. So sánh và tổng hợp kết quả

Kết quả được tổng hợp thành bảng và biểu đồ để so sánh các mô hình trên từng bộ dữ liệu phù hợp. Các mô hình trong cùng một phép thử cần được đánh giá trên cùng tập kiểm tra và cùng biến mục tiêu. Những nhận xét và kết luận chỉ được đưa ra sau khi hoàn thành quá trình thực nghiệm.

## 2.2. Giới thiệu các bộ dữ liệu

Nghiên cứu sử dụng ba bộ dữ liệu có liên quan đến HIV/AIDS. Các bộ dữ liệu được lưu trữ dưới định dạng CSV và được đọc, xử lý bằng Python.

### 2.2.1. Bộ dữ liệu 1: HIV/AIDS prevalence estimates

Bộ dữ liệu `HIV_AIDS prevalence estimates table.csv` chứa các thông tin ước tính về tỷ lệ hiện nhiễm HIV/AIDS ở người trưởng thành, số người sống chung với HIV/AIDS, số ca tử vong hằng năm liên quan đến AIDS và năm ước tính.

### 2.2.2. Bộ dữ liệu 2: AIDS regional aggregates

Bộ dữ liệu `aids_regional_aggregates.csv` cung cấp các chỉ số tổng hợp liên quan đến HIV/AIDS theo quốc gia, khu vực và năm. Các thuộc tính có thể bao gồm dân số, tỷ lệ hiện nhiễm HIV, số người sống chung với HIV, số ca nhiễm mới, số ca tử vong do AIDS và độ bao phủ điều trị ART.

Bộ dữ liệu này có thể hỗ trợ phân tích mối liên hệ giữa tình hình HIV/AIDS với các yếu tố dân số, kinh tế và y tế nếu những thuộc tính tương ứng có mặt và đủ chất lượng.

### 2.2.3. Bộ dữ liệu 3: ART coverage by country

Bộ dữ liệu `art_coverage_by_country_clean.csv` chứa thông tin về số người đang nhận điều trị ART, số người ước tính sống chung với HIV và tỷ lệ bao phủ điều trị ART theo quốc gia hoặc khu vực.

Bộ dữ liệu này được sử dụng để khảo sát sự khác biệt về độ bao phủ điều trị giữa các quốc gia hoặc khu vực, đồng thời xem xét khả năng sử dụng các chỉ số liên quan đến điều trị trong những phân tích tiếp theo.

Việc khảo sát từng bộ dữ liệu riêng biệt giúp xác định cấu trúc, giới hạn và các biến phù hợp trước khi quyết định phương pháp phân tích và dự báo.

```python
from pathlib import Path
import pandas as pd

DATA_DIR = Path("../data/raw")

file_1 = DATA_DIR / "HIV_AIDS prevalence estimates table.csv"
file_2 = DATA_DIR / "aids_regional_aggregates.csv"
file_3 = DATA_DIR / "art_coverage_by_country_clean.csv"

def doc_csv(path):
    if not path.exists():
        raise FileNotFoundError(
            f"Không tìm thấy file: {path.resolve()}"
        )

    for encoding in ["utf-8-sig", "utf-8", "cp1252", "latin1"]:
        try:
            return pd.read_csv(path, encoding=encoding)
        except UnicodeDecodeError:
            continue

df_hiv = doc_csv(file_1)
df_aids = doc_csv(file_2)
df_art = doc_csv(file_3)

print("BỘ DỮ LIỆU 1")
print("Số dòng, số cột:", df_hiv.shape)
display(df_hiv.head())

print("BỘ DỮ LIỆU 2")
print("Số dòng, số cột:", df_aids.shape)
display(df_aids.head())

print("BỘ DỮ LIỆU 3")
print("Số dòng, số cột:", df_art.shape)
display(df_art.head())
```


## 2.4. Làm sạch và tiền xử lý dữ liệu

Quá trình làm sạch dữ liệu được thực hiện nhằm nâng cao chất lượng dữ liệu trước khi phân tích và xây dựng mô hình dự báo. Các công việc bao gồm kiểm tra giá trị thiếu, chuẩn hóa những ký hiệu biểu thị dữ liệu không có sẵn, chuyển đổi kiểu dữ liệu phù hợp và kiểm tra tính hợp lệ của các thuộc tính.

Đối với các cột số liệu, những giá trị được biểu diễn dưới dạng chuỗi sẽ được xem xét chuyển đổi sang kiểu số. Những giá trị không có sẵn như `Nodata` hoặc `-` cần được nhận diện và xử lý phù hợp. Các dòng dữ liệu không bị xóa tùy tiện, nhằm hạn chế mất thông tin có giá trị cho quá trình phân tích.


```python

# Kiểm tra các giá trị đặc biệt trong 3 bộ dữ liệu

datasets = {
    "HIV/AIDS prevalence estimates": df_hiv,
    "AIDS regional aggregates": df_aids,
    "ART coverage by country": df_art
}

for ten, df in datasets.items():
    print("\n" + "=" * 70)
    print("BỘ DỮ LIỆU:", ten)

    for col in df.columns:
        if df[col].dtype == "object" or df[col].dtype == "str":
            gia_tri = df[col].astype("string").str.strip()
            dac_biet = gia_tri[
                gia_tri.str.lower().isin(
                    ["nodata", "no data", "-", "n/a", "na", ""]
                )
            ]

            if not dac_biet.empty:
                print(f"\nCột: {col}")
                print(dac_biet.value_counts().to_dict())

print("\nHoàn tất kiểm tra các giá trị đặc biệt.")
```

```python

import pandas as pd

# Tạo bản sao, giữ nguyên dữ liệu gốc
df_hiv_clean = df_hiv.copy()
df_aids_clean = df_aids.copy()
df_art_clean = df_art.copy()

# Hàm chuẩn hóa dữ liệu số đang được lưu dưới dạng chuỗi
def chuan_hoa_so(series):
    s = series.astype("string").str.strip()

    # Chuẩn hóa các ký hiệu biểu thị dữ liệu không có sẵn
    s = s.replace(
        r"(?i)^\s*(nodata|no data|n/a|na|-)?\s*$",
        pd.NA,
        regex=True
    )

    # Bỏ dấu phẩy phân cách hàng nghìn, ví dụ 11,000
    s = s.str.replace(",", "", regex=False)

    return pd.to_numeric(s, errors="coerce")

# Bộ dữ liệu 1: chuyển các cột số sang kiểu số
cot_so_hiv = [
    "Adult prevalence of HIV/AIDS",
    "Number of people with HIV/AIDS",
    "Annual deaths from HIV/AIDS",
    "Year of estimate"
]

for col in cot_so_hiv:
    df_hiv_clean[col] = chuan_hoa_so(df_hiv_clean[col])

# Bộ dữ liệu 3: chuyển ba cột số liệu chính sang kiểu số
cot_so_art = [
    "Reported number of people receiving ART",
    "Estimated number of people living with HIV",
    "Estimated ART coverage among people living with HIV (%)"
]

for col in cot_so_art:
    df_art_clean[col] = chuan_hoa_so(df_art_clean[col])

# Kiểm tra kết quả chuyển đổi
print("BỘ DỮ LIỆU HIV/AIDS SAU CHUYỂN ĐỔI")
display(df_hiv_clean.dtypes.to_frame("Kiểu dữ liệu"))

print("\nBỘ DỮ LIỆU ART SAU CHUYỂN ĐỔI")
display(df_art_clean.dtypes.to_frame("Kiểu dữ liệu"))
```

```python

# Bước 8.3: Kiểm tra giá trị thiếu trong bộ dữ liệu thứ hai

df_aids_clean = df_aids.copy()

print("KÍCH THƯỚC BAN ĐẦU:", df_aids_clean.shape)

# Thống kê giá trị thiếu
thieu = df_aids_clean.isna().sum()
ty_le_thieu = (thieu / len(df_aids_clean) * 100).round(2)

bao_cao_thieu = pd.DataFrame({
    "So_gia_tri_thieu": thieu,
    "Ty_le_thieu_pct": ty_le_thieu
})

print("\nBÁO CÁO GIÁ TRỊ THIẾU:")
display(
    bao_cao_thieu[bao_cao_thieu["So_gia_tri_thieu"] > 0]
    .sort_values("Ty_le_thieu_pct", ascending=False)
)

# Xác định cột thiếu hoàn toàn
cot_thieu_hoan_toan = [
    col for col in df_aids_clean.columns
    if df_aids_clean[col].isna().all()
]

print("\nCÁC CỘT THIẾU HOÀN TOÀN:")
print(cot_thieu_hoan_toan)

# Kiểm tra một vài dòng dữ liệu
print("\n5 DÒNG ĐẦU TIÊN:")
display(df_aids_clean.head())
```

```python

# Bước 8.5: Làm sạch sơ bộ ba bộ dữ liệu

# 1. Bộ dữ liệu HIV/AIDS prevalence estimates
# Sửa ký tự lỗi ở đầu tên quốc gia nếu có
df_hiv_clean["Country/Region"] = (
    df_hiv_clean["Country/Region"]
    .astype("string")
    .str.replace(r"^ÿ", "", regex=True)
    .str.strip()
)

# 2. Bộ dữ liệu AIDS regional aggregates
# Loại các cột thiếu hoàn toàn khỏi bản dùng để phân tích
cot_thieu_aids = df_aids_clean.columns[
    df_aids_clean.isna().all()
].tolist()

df_aids_clean = df_aids_clean.drop(
    columns=cot_thieu_aids
)

# 3. Bộ dữ liệu ART coverage by country
df_art_clean["Country"] = (
    df_art_clean["Country"]
    .astype("string")
    .str.strip()
)

# Báo cáo các cột đã loại
print("Các cột thiếu hoàn toàn đã loại khỏi bản AIDS:")
print(cot_thieu_aids)

# Kiểm tra kích thước sau làm sạch
print("\nKÍCH THƯỚC SAU LÀM SẠCH SƠ BỘ")

for ten, df in {
    "HIV/AIDS prevalence estimates": df_hiv_clean,
    "AIDS regional aggregates": df_aids_clean,
    "ART coverage by country": df_art_clean
}.items():
    print(f"{ten}: {df.shape}")

# Kiểm tra một số tên quốc gia
print("\nMột số tên trong bộ dữ liệu HIV/AIDS:")
display(df_hiv_clean[["Country/Region"]].head(10))
```

### 3.4. Phân tích tỷ lệ bao phủ điều trị ART

Phần này phân tích tỷ lệ bao phủ điều trị kháng virus (ART) trong số người sống chung với HIV ở các quốc gia có dữ liệu. Sử dụng biến `Estimated ART coverage among people living with HIV (%)_median` để so sánh tỷ lệ bao phủ điều trị giữa các quốc gia, đồng thời khảo sát những quốc gia có tỷ lệ cao nhất và thấp nhất trong bộ dữ liệu.

```python

import matplotlib.pyplot as plt

cot_art = "Estimated ART coverage among people living with HIV (%)_median"

# Lấy 10 quốc gia có tỷ lệ bao phủ ART cao nhất
top10_art = df_art_analysis.nlargest(10, cot_art).sort_values(cot_art)

plt.figure(figsize=(10, 6))

plt.barh(
    top10_art["Country"],
    top10_art[cot_art]
)

plt.title("Top 10 quốc gia có tỷ lệ bao phủ điều trị ART cao nhất")
plt.xlabel("Tỷ lệ bao phủ ART (%)")
plt.ylabel("Quốc gia")
plt.xlim(0, 100)
plt.tight_layout()
plt.show()

# Hiển thị bảng dữ liệu
bang_top10_art = top10_art[
    ["Country", cot_art, "WHO Region"]
].sort_values(cot_art, ascending=False).reset_index(drop=True)

display(bang_top10_art)
```

```python

# Lấy 10 quốc gia có tỷ lệ bao phủ ART thấp nhất
bottom10_art = df_art_analysis.nsmallest(10, cot_art).sort_values(cot_art)

plt.figure(figsize=(10, 6))

plt.barh(
    bottom10_art["Country"],
    bottom10_art[cot_art]
)

plt.title("Top 10 quốc gia có tỷ lệ bao phủ điều trị ART thấp nhất")
plt.xlabel("Tỷ lệ bao phủ ART (%)")
plt.ylabel("Quốc gia")
plt.xlim(0, 100)
plt.tight_layout()
plt.show()

display(
    bottom10_art[
        ["Country", cot_art, "WHO Region"]
    ].reset_index(drop=True)
)
```

- **Số người sống chung với HIV/AIDS:** Dữ liệu có sự chênh lệch lớn giữa các quốc gia và khu vực. Vì vậy, cần xem xét phân bố dữ liệu trước khi lựa chọn phương pháp phân tích hoặc xây dựng mô hình.

- **Số ca tử vong do HIV/AIDS:** Chỉ có 66 quan sát có đầy đủ số ca tử vong và năm ước tính. Kết quả biểu đồ phản ánh các quan sát có dữ liệu, chưa đủ để kết luận về xu hướng tử vong trên toàn thế giới.

- **Tỷ lệ bao phủ điều trị ART:** Có 136 quan sát có dữ liệu về tỷ lệ bao phủ điều trị. Biểu đồ Top 10 quốc gia cao nhất và thấp nhất giúp nhận biết sự khác biệt giữa các quốc gia trong bộ dữ liệu.

- **Chất lượng dữ liệu:** Cả ba bộ dữ liệu đều còn giá trị thiếu. Do đó, cần lựa chọn biến mục tiêu, kiểm tra kiểu dữ liệu và xử lý dữ liệu thiếu phù hợp trước khi xây dựng mô hình.

Nhìn chung, EDA giúp nhận diện đặc điểm, sự chênh lệch và những hạn chế của dữ liệu HIV/AIDS. Các kết quả này là cơ sở để lựa chọn biến phân tích và chuẩn bị dữ liệu cho giai đoạn xây dựng, đánh giá mô hình tiếp theo.
