# CHƯƠNG 1. TỔNG QUAN ĐỀ TÀI

## 1.1. Lý do chọn đề tài

HIV/AIDS là một trong những vấn đề sức khỏe cộng đồng nhận được sự quan tâm trên phạm vi toàn cầu. HIV là virus gây suy giảm miễn dịch ở người, trong khi AIDS là hội chứng suy giảm miễn dịch mắc phải có thể xuất hiện ở giai đoạn tiến triển của tình trạng nhiễm HIV nếu người bệnh không được điều trị hiệu quả.

Mặc dù đã có nhiều tiến bộ trong công tác phòng ngừa, xét nghiệm và điều trị, HIV/AIDS vẫn đặt ra nhiều thách thức đối với hệ thống y tế. Việc theo dõi số người sống chung với HIV, số ca nhiễm mới, số ca tử vong liên quan đến AIDS và tỷ lệ bao phủ điều trị bằng thuốc kháng virus có vai trò quan trọng trong việc tìm hiểu diễn biến của dịch bệnh.

Các chỉ số HIV/AIDS có thể khác nhau giữa các quốc gia, khu vực và thời điểm. Vì vậy, việc thu thập, xử lý và phân tích dữ liệu là cần thiết nhằm nhận diện xu hướng, tìm hiểu đặc điểm của dữ liệu và hỗ trợ đánh giá tình hình dịch bệnh.

Sự phát triển của công nghệ thông tin và học máy tạo điều kiện ứng dụng các phương pháp phân tích dữ liệu vào những vấn đề thực tiễn. Với ngôn ngữ Python và các thư viện như Pandas, NumPy, Matplotlib và Scikit-learn, người nghiên cứu có thể thực hiện quy trình từ tiền xử lý dữ liệu, phân tích khám phá đến xây dựng và đánh giá mô hình dự báo.

Xuất phát từ những lý do trên, đề tài “Phân tích dữ liệu HIV/AIDS toàn cầu và so sánh các mô hình hồi quy trong dự báo các chỉ số HIV/AIDS” được thực hiện nhằm khảo sát ba bộ dữ liệu liên quan đến HIV/AIDS, phân tích các đặc điểm nổi bật và so sánh hiệu quả của năm mô hình hồi quy trên những dữ liệu phù hợp. Qua đó, đề tài vận dụng kiến thức lập trình Python, phân tích dữ liệu và học máy vào một vấn đề có ý nghĩa đối với sức khỏe cộng đồng.

## 1.2. Mục tiêu nghiên cứu

### 1.2.1. Mục tiêu tổng quát

Phân tích ba bộ dữ liệu liên quan đến HIV/AIDS, khám phá đặc điểm và xu hướng của các chỉ số quan trọng, đồng thời xây dựng và so sánh năm mô hình hồi quy nhằm đánh giá khả năng dự báo các chỉ số định lượng phù hợp với dữ liệu.

### 1.2.2. Mục tiêu cụ thể

- Khảo sát cấu trúc, nội dung và chất lượng của ba bộ dữ liệu.
- Tiền xử lý dữ liệu, kiểm tra giá trị thiếu, dữ liệu trùng lặp và những sai lệch về định dạng.
- Thực hiện thống kê mô tả và trực quan hóa các chỉ số HIV/AIDS.

- Xây dựng năm mô hình hồi quy: Linear Regression, Ridge Regression, Random Forest Regressor, Gradient Boosting Regressor và Support Vector Regression (SVR).
- Đánh giá và so sánh mô hình bằng các chỉ số MAE, RMSE và R².

## 1.5. Phương pháp nghiên cứu

### 1.5.1. Khảo sát dữ liệu

Tiến hành đọc và kiểm tra ba bộ dữ liệu, xác định số lượng dòng, số lượng cột, tên thuộc tính, kiểu dữ liệu và phạm vi quan sát. Đồng thời, đánh giá mức độ phù hợp của từng bộ dữ liệu đối với mục tiêu phân tích và dự báo.

### 1.5.2. Tiền xử lý dữ liệu

Kiểm tra giá trị thiếu, bản ghi trùng lặp, dữ liệu không hợp lệ và sai lệch định dạng. Thực hiện chuyển đổi kiểu dữ liệu, xử lý giá trị thiếu và mã hóa biến phân loại khi cần thiết. Những thao tác tiền xử lý được lựa chọn dựa trên đặc điểm của từng bộ dữ liệu nhằm hạn chế làm sai lệch thông tin.

### 1.5.3. Phân tích khám phá dữ liệu

Sử dụng thống kê mô tả và biểu đồ để tìm hiểu sự phân bố của các chỉ số HIV/AIDS, sự thay đổi theo thời gian và sự khác biệt giữa các nhóm quan sát khi dữ liệu cho phép. Kết quả được trình bày bằng bảng thống kê và hình ảnh trực quan.

### 1.5.4. Xây dựng mô hình hồi quy

Đề tài dự kiến sử dụng năm mô hình:

1. Linear Regression: mô hình hồi quy tuyến tính cơ sở.
2. Ridge Regression: hồi quy tuyến tính có điều chuẩn L2.
3. Random Forest Regressor: mô hình tổ hợp nhiều cây quyết định.
4. Gradient Boosting Regressor: mô hình tăng cường tuần tự.
5. Support Vector Regression (SVR): mô hình hồi quy dựa trên véc-tơ hỗ trợ.

Việc lựa chọn biến mục tiêu và thiết lập mô hình được thực hiện dựa trên cấu trúc dữ liệu thực tế. Các mô hình được huấn luyện và đánh giá theo quy trình phù hợp nhằm hạn chế rò rỉ dữ liệu.

### 1.5.5. Đánh giá mô hình

Các mô hình được đánh giá bằng ba chỉ số:

- MAE: đo sai số tuyệt đối trung bình; giá trị thấp hơn thường thể hiện dự báo gần giá trị thực tế hơn.
- RMSE: đo căn bậc hai của sai số bình phương trung bình; nhạy hơn với các sai số lớn.
- R²: phản ánh mức độ phù hợp của dự báo so với mức tham chiếu dựa trên giá trị trung bình của biến mục tiêu.

Việc lựa chọn mô hình phù hợp căn cứ vào kết quả thực nghiệm, các chỉ số đánh giá và đặc điểm của từng bộ dữ liệu.

### 1.5.6. So sánh và tổng hợp kết quả

Kết quả được tổng hợp thành bảng và biểu đồ để so sánh các mô hình trên từng bộ dữ liệu phù hợp. Các mô hình trong cùng một phép thử cần được đánh giá trên cùng tập kiểm tra và cùng biến mục tiêu. Những nhận xét và kết luận chỉ được đưa ra sau khi hoàn thành quá trình thực nghiệm.

## 2.1. Giới thiệu chương

Nghiên cứu sử dụng ba bộ dữ liệu liên quan đến tình hình HIV/AIDS, các chỉ số tổng hợp theo quốc gia và khu vực, cùng thông tin về độ bao phủ điều trị bằng thuốc kháng vi-rút (ART). Mỗi bộ dữ liệu cung cấp những thông tin khác nhau, góp phần hỗ trợ việc tìm hiểu tình hình HIV/AIDS và lựa chọn các biến phù hợp cho quá trình phân tích.

Các bước thực hiện bao gồm: giới thiệu dữ liệu, kiểm tra cấu trúc và kiểu dữ liệu, đánh giá giá trị thiếu và dữ liệu trùng lặp, xử lý những vấn đề phát hiện được, sau đó chuẩn bị dữ liệu phù hợp cho các bước phân tích tiếp theo.

Kết quả của chương này là cơ sở để thực hiện phân tích khám phá dữ liệu, lựa chọn biến đầu vào và biến mục tiêu, cũng như xây dựng các mô hình dự báo ở những chương sau.

## 2.2. Giới thiệu các bộ dữ liệu

Nghiên cứu sử dụng ba bộ dữ liệu có liên quan đến HIV/AIDS. Các bộ dữ liệu được lưu trữ dưới định dạng CSV và được đọc, xử lý bằng Python.

### 2.2.1. Bộ dữ liệu 1: HIV/AIDS prevalence estimates

Bộ dữ liệu `HIV_AIDS prevalence estimates table.csv` chứa các thông tin ước tính về tỷ lệ hiện nhiễm HIV/AIDS ở người trưởng thành, số người sống chung với HIV/AIDS, số ca tử vong hằng năm liên quan đến AIDS và năm ước tính.

Bộ dữ liệu này được sử dụng để khảo sát sự khác biệt về tình hình HIV/AIDS giữa các quốc gia hoặc khu vực theo những thông tin có trong dữ liệu.

### 2.2.2. Bộ dữ liệu 2: AIDS regional aggregates

Bộ dữ liệu `aids_regional_aggregates.csv` gồm 455 dòng và 26 cột, là chuỗi thời gian 1990–2024 của 13 nhóm khu vực hoặc nhóm thu nhập (theo cách phân nhóm của World Bank, ví dụ *Sub-Saharan Africa*), **không phải từng quốc gia**. Các thuộc tính có dữ liệu gồm dân số, GDP bình quân đầu người, tuổi thọ, chi tiêu y tế, tỷ lệ dân thành thị, tỷ lệ nghèo và `HIV_Prevalence_WB`.

13 cột chỉ số HIV/AIDS (số người sống chung với HIV, tỷ lệ hiện nhiễm, số ca nhiễm mới, số ca tử vong do AIDS, độ bao phủ ART, ...) **không có dữ liệu** nên bị loại ở bước làm sạch. Chỉ `HIV_Prevalence_WB` còn giá trị (107/455 dòng, chủ yếu thuộc 3 vùng), vì vậy bộ dữ liệu này được dùng cho bài toán dự báo theo thời gian ở Chương 4.

### 2.2.3. Bộ dữ liệu 3: ART coverage by country

Bộ dữ liệu `art_coverage_by_country_clean.csv` chứa thông tin về số người đang nhận điều trị ART, số người ước tính sống chung với HIV và tỷ lệ bao phủ điều trị ART theo quốc gia hoặc khu vực.

Bộ dữ liệu này được sử dụng để khảo sát sự khác biệt về độ bao phủ điều trị giữa các quốc gia hoặc khu vực, đồng thời xem xét khả năng sử dụng các chỉ số liên quan đến điều trị trong những phân tích tiếp theo.

## 2.4. Làm sạch và tiền xử lý dữ liệu

Quá trình làm sạch dữ liệu được thực hiện nhằm nâng cao chất lượng dữ liệu trước khi phân tích và xây dựng mô hình dự báo. Các công việc bao gồm kiểm tra giá trị thiếu, chuẩn hóa những ký hiệu biểu thị dữ liệu không có sẵn, chuyển đổi kiểu dữ liệu phù hợp và kiểm tra tính hợp lệ của các thuộc tính.

Đối với các cột số liệu, những giá trị được biểu diễn dưới dạng chuỗi sẽ được xem xét chuyển đổi sang kiểu số. Những giá trị không có sẵn như `Nodata` hoặc `-` cần được nhận diện và xử lý phù hợp. Các dòng dữ liệu không bị xóa tùy tiện, nhằm hạn chế mất thông tin có giá trị cho quá trình phân tích.

```python
# Kiểm tra các giá trị đặc biệt trong 3 bộ dữ liệu

datasets = {
    "HIV/AIDS prevalence estimates": df_hiv,
    "AIDS regional aggregates": df_aids,
    "ART coverage by country": df_art
}

for ten, df in datasets.items():
    print("\n" + "=" * 70)
    print("BỘ DỮ LIỆU:", ten)

    for col in df.columns:
        if df[col].dtype == "object" or df[col].dtype == "str":
            gia_tri = df[col].astype("string").str.strip()
            dac_biet = gia_tri[
                gia_tri.str.lower().isin(
                    ["nodata", "no data", "-", "n/a", "na", ""]
                )
            ]

            if not dac_biet.empty:
                print(f"\nCột: {col}")
                print(dac_biet.value_counts().to_dict())

print("\nHoàn tất kiểm tra các giá trị đặc biệt.")
```

```python
import pandas as pd

# Tạo bản sao, giữ nguyên dữ liệu gốc
df_hiv_clean = df_hiv.copy()
df_aids_clean = df_aids.copy()
df_art_clean = df_art.copy()

# Hàm chuẩn hóa dữ liệu số đang được lưu dưới dạng chuỗi
def chuan_hoa_so(series):
    """'28.30%' -> 28.3 ; '1,400,000[11]' -> 1400000 ; '1 265' -> 1265 ; '-' / 'Nodata' -> NA"""
    s = series.astype("string")
    s = s.str.replace(r"\[.*?\]", "", regex=True)            # bỏ chú thích [10], [unreliable source]
    s = s.str.replace(r"\s+", "", regex=True)                 # bỏ mọi khoảng trắng (kể cả \xa0 làm dấu phân cách nghìn)
    s = s.str.replace("%", "", regex=False).str.replace(",", "", regex=False)
    s = s.replace(r"(?i)^(nodata|no data|n/a|na|-)?$", pd.NA, regex=True)
    return pd.to_numeric(s, errors="coerce")

def chuan_hoa_ten(series):
    """Bỏ chú thích [..], ký tự khoảng trắng đặc biệt và khoảng trắng thừa trong tên quốc gia."""
    s = series.astype("string").str.replace(r"\[.*?\]", "", regex=True)
    return s.str.replace(r"\s+", " ", regex=True).str.strip()

# Bộ dữ liệu 1: chuyển các cột số sang kiểu số
cot_so_hiv = [
    "Adult prevalence of HIV/AIDS",
    "Number of people with HIV/AIDS",
    "Annual deaths from HIV/AIDS",
    "Year of estimate"
]

for col in cot_so_hiv:
    df_hiv_clean[col] = chuan_hoa_so(df_hiv_clean[col])

# Bộ dữ liệu 3: chuyển ba cột số liệu chính sang kiểu số
cot_so_art = [
    "Reported number of people receiving ART",
    "Estimated number of people living with HIV",
    "Estimated ART coverage among people living with HIV (%)"
]

for col in cot_so_art:
    df_art_clean[col] = chuan_hoa_so(df_art_clean[col])

# Kiểm tra kết quả chuyển đổi
print("BỘ DỮ LIỆU HIV/AIDS SAU CHUYỂN ĐỔI")
display(df_hiv_clean.dtypes.to_frame("Kiểu dữ liệu"))

print("\nBỘ DỮ LIỆU ART SAU CHUYỂN ĐỔI")
display(df_art_clean.dtypes.to_frame("Kiểu dữ liệu"))
```

```python
# Bước 8.3: Kiểm tra giá trị thiếu trong bộ dữ liệu thứ hai

df_aids_clean = df_aids.copy()

print("KÍCH THƯỚC BAN ĐẦU:", df_aids_clean.shape)

# Thống kê giá trị thiếu
thieu = df_aids_clean.isna().sum()
ty_le_thieu = (thieu / len(df_aids_clean) * 100).round(2)

bao_cao_thieu = pd.DataFrame({
    "So_gia_tri_thieu": thieu,
    "Ty_le_thieu_pct": ty_le_thieu
})

print("\nBÁO CÁO GIÁ TRỊ THIẾU:")
display(
    bao_cao_thieu[bao_cao_thieu["So_gia_tri_thieu"] > 0]
    .sort_values("Ty_le_thieu_pct", ascending=False)
)

# Xác định cột thiếu hoàn toàn
cot_thieu_hoan_toan = [
    col for col in df_aids_clean.columns
    if df_aids_clean[col].isna().all()
]

print("\nCÁC CỘT THIẾU HOÀN TOÀN:")
print(cot_thieu_hoan_toan)

# Kiểm tra một vài dòng dữ liệu
print("\n5 DÒNG ĐẦU TIÊN:")
display(df_aids_clean.head())
```

```python
# Bước 8.5: Làm sạch sơ bộ ba bộ dữ liệu

# 1. Bộ dữ liệu HIV/AIDS prevalence estimates
# Sửa ký tự lỗi ở đầu tên quốc gia nếu có
df_hiv_clean["Country/Region"] = chuan_hoa_ten(df_hiv_clean["Country/Region"])

# 2. Bộ dữ liệu AIDS regional aggregates
# Loại các cột thiếu hoàn toàn khỏi bản dùng để phân tích
cot_thieu_aids = df_aids_clean.columns[
    df_aids_clean.isna().all()
].tolist()

df_aids_clean = df_aids_clean.drop(
    columns=cot_thieu_aids
)

# 3. Bộ dữ liệu ART coverage by country
df_art_clean["Country"] = (
    df_art_clean["Country"]
    .astype("string")
    .str.strip()
)

# Báo cáo các cột đã loại
print("Các cột thiếu hoàn toàn đã loại khỏi bản AIDS:")
print(cot_thieu_aids)

# Kiểm tra kích thước sau làm sạch
print("\nKÍCH THƯỚC SAU LÀM SẠCH SƠ BỘ")

for ten, df in {
    "HIV/AIDS prevalence estimates": df_hiv_clean,
    "AIDS regional aggregates": df_aids_clean,
    "ART coverage by country": df_art_clean
}.items():
    print(f"{ten}: {df.shape}")

# Kiểm tra một số tên quốc gia
print("\nMột số tên trong bộ dữ liệu HIV/AIDS:")
display(df_hiv_clean[["Country/Region"]].head(10))
```

## 4.1. Thư viện, đọc và làm sạch dữ liệu

Mục này tự đọc lại dữ liệu gốc bằng các hàm làm sạch đã sửa lỗi (bỏ `%`, chú thích `[..]`, khoảng trắng phân cách nghìn), nên chạy độc lập với Chương 2.

```python
import re
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.dummy import DummyRegressor
from sklearn.model_selection import train_test_split, KFold, TimeSeriesSplit, cross_validate
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

sns.set_theme(style="whitegrid")
# --- Tiện ích chạy ổn định ở mọi môi trường ---
try:
    display                      # có sẵn trong Jupyter / VS Code
except NameError:                # chạy bằng "python file.py" thì không có display
    def display(x):
        print(x.to_string() if hasattr(x, "to_string") else x)

def tim_thu_muc_goc():
    """Tìm thư mục dự án (nơi có data/raw), dù chạy từ notebooks/, từ thư mục gốc hay bằng python file.py."""
    ung_vien = [Path.cwd().resolve()]
    if "__file__" in globals():
        ung_vien.append(Path(__file__).resolve().parent)
    for p in ung_vien:
        for q in [p, *p.parents]:
            if (q / "data" / "raw").is_dir():
                return q
    raise FileNotFoundError("Không tìm thấy thư mục data/raw. Hãy chạy từ bên trong thư mục dự án DH23TIN07_LT_PYTHON_HIV.")

ROOT = tim_thu_muc_goc()

RAW = ROOT / "data" / "raw"
PROCESSED = ROOT / "data" / "processed"
PROCESSED.mkdir(parents=True, exist_ok=True)
RANDOM_STATE = 42
```

```python
def lam_sach_ten(s):
    s = re.sub(r"\[.*?\]", "", str(s))              # bỏ chú thích [10]
    return re.sub(r"\s+", " ", s).strip()

def sang_so(s):
    """'28.30%' -> 28.3 ; '1,400,000[11]' -> 1400000 ; '1\\xa0265' -> 1265 ; '-' -> NaN"""
    if pd.isna(s):
        return np.nan
    s = re.sub(r"\[.*?\]", "", str(s))
    s = re.sub(r"\s+", "", s).replace("%", "").replace(",", "")
    try:
        return float(s)
    except ValueError:
        return np.nan

# ---- Bảng 1: prevalence (file không phải UTF-8 nên đọc bằng cp850) ----
prev_raw = pd.read_csv(RAW / "HIV_AIDS prevalence estimates table.csv", encoding="cp850")
df_prev = pd.DataFrame({
    "country":        prev_raw["Country/Region"].map(lam_sach_ten),
    "prevalence_pct": prev_raw["Adult prevalence of HIV/AIDS"].map(sang_so),
    "plhiv":          prev_raw["Number of people with HIV/AIDS"].map(sang_so),
    "deaths":         prev_raw["Annual deaths from HIV/AIDS"].map(sang_so),
    "est_year":       prev_raw["Year of estimate"].map(sang_so),
})

# ---- Bảng 2: regional (bỏ cột trống 100%, cột trùng/hằng số và nhóm không có dữ liệu) ----
reg_raw = pd.read_csv(RAW / "aids_regional_aggregates.csv")
cot_rong = reg_raw.columns[reg_raw.isna().all()].tolist()
df_reg = (reg_raw.drop(columns=cot_rong + ["ISO_Code", "Source"])
                 .rename(columns={"Country": "Region"}))
df_reg = df_reg[df_reg["Region"] != "Not classified"].reset_index(drop=True)

# ---- Bảng 3: ART ('Nodata' -> NaN, dùng các cột median dạng số) ----
art_raw = pd.read_csv(RAW / "art_coverage_by_country_clean.csv")
df_art_mh = art_raw.replace("Nodata", np.nan).rename(columns={
    "Country": "country",
    "Reported number of people receiving ART": "art_reported",
    "Estimated number of people living with HIV_median": "plhiv_median",
    "Estimated number of people living with HIV_min": "plhiv_min",
    "Estimated number of people living with HIV_max": "plhiv_max",
    "Estimated ART coverage among people living with HIV (%)_median": "art_cov_median",
    "Estimated ART coverage among people living with HIV (%)_min": "art_cov_min",
    "Estimated ART coverage among people living with HIV (%)_max": "art_cov_max",
    "WHO Region": "who_region",
})
df_art_mh["art_reported"] = pd.to_numeric(df_art_mh["art_reported"], errors="coerce")
df_art_mh = df_art_mh.drop(columns=[c for c in df_art_mh.columns if "[" in str(c) or c.startswith("Estimated")])

print("Số giá trị hợp lệ - Bảng 1:", df_prev.notna().sum().to_dict())
print("Bảng 2:", df_reg.shape, "| dòng có HIV_Prevalence_WB:", int(df_reg["HIV_Prevalence_WB"].notna().sum()))
print("Số giá trị hợp lệ - Bảng 3:", df_art_mh.notna().sum().to_dict())
```

## 4.2. Thiết lập chung cho mọi bài toán

| Mô hình | Thiết lập | Ghi chú |
|---|---|---|
| Linear Regression | mặc định | mô hình tuyến tính cơ sở |
| Ridge Regression | `alpha=1.0` | điều chuẩn L2, giảm ảnh hưởng của đa cộng tuyến |
| Random Forest | 300 cây, `min_samples_leaf=2` | không ngoại suy được ngoài miền giá trị đã học |
| Gradient Boosting | 200 cây, `learning_rate=0.05`, `max_depth=3` | tăng cường tuần tự |
| SVR | kernel RBF, `C=10`, chuẩn hóa biến mục tiêu | nhạy với thang đo |
| Baseline | luôn dự đoán trung bình tập huấn luyện | mốc tham chiếu (R² ≈ 0) |

Biến số được chuẩn hóa, biến phân loại được mã hóa one-hot, tất cả nằm trong `Pipeline` nên chỉ "học" từ tập huấn luyện (hạn chế rò rỉ dữ liệu).

```python
def tao_mo_hinh():
    return {
        "Baseline (trung bình)": DummyRegressor(strategy="mean"),
        "Linear Regression": LinearRegression(),
        "Ridge Regression": Ridge(alpha=1.0),
        "Random Forest": RandomForestRegressor(n_estimators=300, min_samples_leaf=2,
                                               random_state=RANDOM_STATE),
        "Gradient Boosting": GradientBoostingRegressor(n_estimators=200, learning_rate=0.05,
                                                       max_depth=3, random_state=RANDOM_STATE),
        "SVR": TransformedTargetRegressor(regressor=SVR(kernel="rbf", C=10, epsilon=0.1),
                                          transformer=StandardScaler()),
    }

def tao_pipeline(model, cot_so, cot_phan_loai):
    bo_tien_xu_ly = ColumnTransformer([
        ("so", StandardScaler(), cot_so),
        ("pl", OneHotEncoder(handle_unknown="ignore"), cot_phan_loai),
    ])
    return Pipeline([("tien_xu_ly", bo_tien_xu_ly), ("mo_hinh", model)])

def rmse(y_that, y_dubao):
    return np.sqrt(mean_squared_error(y_that, y_dubao))

def chay_so_sanh(X_train, y_train, X_test, y_test, cot_so, cot_phan_loai, cv):
    """Huấn luyện từng mô hình; trả về bảng kết quả, dự báo trên test và các pipeline đã học."""
    hang, du_bao, da_hoc = [], {}, {}
    for ten, mo_hinh in tao_mo_hinh().items():
        pipe = tao_pipeline(mo_hinh, cot_so, cot_phan_loai)
        pipe.fit(X_train, y_train)
        y_db = pipe.predict(X_test)
        cv_kq = cross_validate(pipe, X_train, y_train, cv=cv,
                               scoring={"mae": "neg_mean_absolute_error",
                                        "rmse": "neg_root_mean_squared_error",
                                        "r2": "r2"})
        hang.append({
            "Mô hình": ten,
            "MAE (test)": mean_absolute_error(y_test, y_db),
            "RMSE (test)": rmse(y_test, y_db),
            "R² (test)": r2_score(y_test, y_db),
            "MAE (CV)": -cv_kq["test_mae"].mean(),
            "RMSE (CV)": -cv_kq["test_rmse"].mean(),
            "R² (CV)": cv_kq["test_r2"].mean(),
        })
        du_bao[ten], da_hoc[ten] = y_db, pipe
    return pd.DataFrame(hang).set_index("Mô hình"), du_bao, da_hoc

def ve_so_sanh(bang, tieu_de):
    fig, ax = plt.subplots(1, 3, figsize=(15, 4))
    for a, cot in zip(ax, ["MAE (test)", "RMSE (test)", "R² (test)"]):
        bang[cot].plot.barh(ax=a, color="#4C72B0")
        a.set_title(cot); a.set_ylabel("")
    fig.suptitle(tieu_de, y=1.03)
    plt.tight_layout(); plt.show()

def ve_thuc_te_du_bao(y_that, y_db, tieu_de, nhan, log10=False):
    """Biểu đồ thực tế so với dự báo; log10=True thì đưa về thang gốc và vẽ trục log."""
    y_that, y_db = np.asarray(y_that, float), np.asarray(y_db, float)
    if log10:
        y_that, y_db = 10 ** y_that, 10 ** y_db
    fig, ax = plt.subplots(figsize=(5.5, 5.5))
    ax.scatter(y_that, y_db, s=45, alpha=0.8)
    lo, hi = min(y_that.min(), y_db.min()), max(y_that.max(), y_db.max())
    ax.plot([lo, hi], [lo, hi], "r--", label="dự báo hoàn hảo")
    if log10:
        ax.set_xscale("log"); ax.set_yscale("log")
    ax.set_xlabel(f"{nhan} (thực tế)"); ax.set_ylabel(f"{nhan} (dự báo)")
    ax.set_title(tieu_de); ax.legend(); plt.tight_layout(); plt.show()
```

## 4.3. Bảng 1 – HIV/AIDS prevalence estimates: dự báo số ca tử vong hằng năm

### 4.3.1. Chuẩn bị dữ liệu

- Mục tiêu: `log10(deaths)`. Số ca tử vong chênh lệch rất lớn giữa các quốc gia (từ vài trăm đến 80.000), nên lấy log10 để phân bố bớt lệch.
- Đầu vào: `log10(plhiv)` (số người sống chung với HIV) và `est_year` (năm ước tính, vì các quốc gia được ước tính ở các năm khác nhau).
- Chỉ giữ quốc gia có đủ cả ba giá trị. Cột `deaths` thiếu 126/193 dòng nên **cỡ mẫu chỉ còn khoảng 67**, đây là hạn chế chính của bài toán này.
- Không đưa `prevalence_pct` vào đầu vào vì làm giảm thêm cỡ mẫu mà không cải thiện kết quả khi thử nghiệm.

```python
t1 = df_prev.dropna(subset=["deaths", "plhiv", "est_year"]).copy()
t1["log_deaths"] = np.log10(t1["deaths"])
t1["log_plhiv"] = np.log10(t1["plhiv"])
t1["est_year"] = t1["est_year"].astype(float)

COT_SO_T1, COT_PL_T1, MUC_TIEU_T1 = ["log_plhiv", "est_year"], [], "log_deaths"
X_train_t1, X_test_t1, y_train_t1, y_test_t1 = train_test_split(
    t1[COT_SO_T1], t1[MUC_TIEU_T1], test_size=0.2, random_state=RANDOM_STATE)

print("Số quốc gia dùng được:", len(t1), "| Train:", len(X_train_t1), "| Test:", len(X_test_t1))
t1[["deaths", "plhiv", "est_year"]].describe().round(1)
```

### 4.3.2. Huấn luyện và đánh giá

Tập test chỉ có khoảng 14 quốc gia, nên cột `(CV)` (5 lần gập trên tập huấn luyện) đáng tin hơn cột `(test)`. Mục 5.5 sẽ lặp lại đánh giá nhiều lần để kiểm tra độ ổn định.

```python
cv_t1 = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
bang_t1, du_bao_t1, mo_hinh_t1 = chay_so_sanh(X_train_t1, y_train_t1, X_test_t1, y_test_t1,
                                              COT_SO_T1, COT_PL_T1, cv_t1)
bang_t1.to_csv(PROCESSED / "ketqua_T1.csv")
bang_t1.sort_values("RMSE (test)").round(3)
```

```python
ve_so_sanh(bang_t1, "Bảng 1 – so sánh 5 mô hình (mục tiêu: log10 số ca tử vong)")
tot_nhat_t1 = bang_t1.drop(index="Baseline (trung bình)")["RMSE (test)"].idxmin()
print("Mô hình tốt nhất trên tập test:", tot_nhat_t1)
ve_thuc_te_du_bao(y_test_t1, du_bao_t1[tot_nhat_t1], f"Bảng 1 – {tot_nhat_t1}",
                  "Số ca tử vong hằng năm", log10=True)
```

## 4.4. Bảng 2 – AIDS regional aggregates: dự báo tỷ lệ hiện nhiễm theo vùng và thời gian

### 4.4.1. Chuẩn bị dữ liệu

- Biến mục tiêu duy nhất còn dữ liệu là `HIV_Prevalence_WB`. Chỉ giữ **3 vùng có đủ 35 năm** (Sub-Saharan Africa, Latin America & Caribbean, Pre-demographic dividend), tổng 105 quan sát.
- Đầu vào: `Year`, `GDP_per_Capita`, `Life_Expectancy`, `Urban_Population_Pct`, `log(Population)` và `Region` (one-hot). Loại `Health_Expenditure` và `Poverty_Headcount_Ratio` vì thiếu nhiều và thiếu theo vùng.
- **Chia theo thời gian** (train ≤ 2016, test 2017–2024), không dùng `shuffle`.

```python
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import RepeatedKFold, cross_validate

sns.set_theme(style="whitegrid")
# --- Tiện ích chạy ổn định ở mọi môi trường ---
try:
    display                      # có sẵn trong Jupyter / VS Code
except NameError:                # chạy bằng "python file.py" thì không có display
    def display(x):
        print(x.to_string() if hasattr(x, "to_string") else x)

def tim_thu_muc_goc():
    """Tìm thư mục dự án (nơi có data/raw), dù chạy từ notebooks/, từ thư mục gốc hay bằng python file.py."""
    ung_vien = [Path.cwd().resolve()]
    if "__file__" in globals():
        ung_vien.append(Path(__file__).resolve().parent)
    for p in ung_vien:
        for q in [p, *p.parents]:
            if (q / "data" / "raw").is_dir():
                return q
    raise FileNotFoundError("Không tìm thấy thư mục data/raw. Hãy chạy từ bên trong thư mục dự án DH23TIN07_LT_PYTHON_HIV.")

ROOT = tim_thu_muc_goc()

PROCESSED = ROOT / "data" / "processed"
FIG = ROOT / "reports" / "figures"      # nơi lưu hình để chèn vào báo cáo
FIG.mkdir(parents=True, exist_ok=True)

TEN_BAI = {
    "T1":  "Bảng 1 (tử vong)",
    "T2":  "Bảng 2 (prevalence theo vùng)",
    "T3":  "Bảng 3 (số người sống chung)",
    "T3b": "Bảng 3b (tỷ lệ bao phủ ART)",
}
TEN_NGAN = {"T1": "Bảng 1", "T2": "Bảng 2", "T3": "Bảng 3", "T3b": "Bảng 3b"}
ket_qua = {k: pd.read_csv(PROCESSED / f"ketqua_{k}.csv", index_col=0) for k in TEN_BAI}

BASELINE = "Baseline (trung bình)"
MO_HINH = [m for m in ket_qua["T1"].index if m != BASELINE]
for k, v in ket_qua.items():
    print(TEN_BAI[k], "->", v.shape)
```

## 5.2. Bảng tổng hợp và xếp hạng

Xếp hạng theo **RMSE** (thấp hơn là tốt hơn), riêng trên tập test và trên CV, rồi lấy hạng trung bình. Hạng trung bình giữa test và CV ổn định hơn việc chỉ nhìn một lần chia.

```python
bang_hang = {}
for k, bang in ket_qua.items():
    b = bang.loc[MO_HINH]
    bang_hang[k] = pd.DataFrame({
        "RMSE (test)": b["RMSE (test)"],
        "Hạng test": b["RMSE (test)"].rank().astype(int),
        "RMSE (CV)": b["RMSE (CV)"],
        "Hạng CV": b["RMSE (CV)"].rank().astype(int),
    }).assign(**{"Hạng TB": lambda d: (d["Hạng test"] + d["Hạng CV"]) / 2}).sort_values("Hạng TB")

for k, b in bang_hang.items():
    print(f"\n=== {TEN_BAI[k]} ===")
    display(b.round(3))
```

```python
# Hạng trung bình của mỗi mô hình qua các bài toán
hang_tong = pd.DataFrame({TEN_BAI[k]: bang_hang[k]["Hạng TB"] for k in TEN_BAI})
hang_tong["Hạng TB chung"] = hang_tong.mean(axis=1)
hang_tong.sort_values("Hạng TB chung").round(2)
```

## 5.3. Mức cải thiện so với mô hình cơ sở

Cải thiện = `1 − RMSE_mô_hình / RMSE_baseline`. Giá trị dương nghĩa là tốt hơn việc luôn dự đoán bằng trung bình; giá trị âm nghĩa là **kém hơn** baseline.

```python
def cai_thien(cot):
    return pd.DataFrame({
        TEN_NGAN[k]: 1 - bang.loc[MO_HINH, cot] / bang.loc[BASELINE, cot]
        for k, bang in ket_qua.items()
    })

ct_test, ct_cv = cai_thien("RMSE (test)"), cai_thien("RMSE (CV)")

fig, ax = plt.subplots(1, 2, figsize=(13, 4.2))
for a, d, tieu_de in [(ax[0], ct_test, "Trên tập test"), (ax[1], ct_cv, "Trên CV")]:
    sns.heatmap(d * 100, annot=True, fmt=".0f", cmap="RdYlGn", center=0, vmin=-60, vmax=100,
                cbar_kws={"label": "% giảm RMSE"}, ax=a)
    a.set_title(f"Giảm RMSE so với baseline – {tieu_de}", fontsize=11)
    a.set_ylabel("")
plt.tight_layout()
plt.savefig(FIG / "hinh_5_1_cai_thien_so_voi_baseline.png", dpi=150, bbox_inches="tight")
plt.show()
```

**Cách đọc:** ô xanh đậm là mô hình giảm sai số nhiều so với baseline, ô đỏ là mô hình tệ hơn baseline. Với Bảng 2, baseline trung bình rất yếu (ba vùng có mức HIV khác nhau nhiều) nên mức cải thiện cao là dễ hiểu; cần đối chiếu thêm mốc naive ở mục 5.5.2. Bài toán nào có hầu hết ô gần 0 hoặc âm thì **không dự báo được** với các đặc trưng hiện có.

## 5.4. Kiểm tra quá khớp: so sánh sai số test với sai số CV

Tỷ số `RMSE (test) / RMSE (CV)` gần 1 nghĩa là mô hình hoạt động ổn định giữa hai cách đánh giá. Tỷ số rất khác 1 là dấu hiệu quá khớp, hoặc quy luật thay đổi giữa giai đoạn train và test, hoặc tập test quá nhỏ nên kết quả dao động.

```python
ty_so = pd.DataFrame({
    TEN_NGAN[k]: bang.loc[MO_HINH, "RMSE (test)"] / bang.loc[MO_HINH, "RMSE (CV)"]
    for k, bang in ket_qua.items()
})
display(ty_so.round(2))

ax = ty_so.plot.bar(figsize=(10, 4), rot=20)
ax.axhline(1, color="gray", ls="--"); ax.set_ylabel("RMSE test / RMSE CV")
ax.set_title("Tỷ số sai số test so với CV (gần 1 là ổn định)")
plt.tight_layout()
plt.savefig(FIG / "hinh_5_2_ty_so_test_cv.png", dpi=150, bbox_inches="tight")
plt.show()
```

### 5.5.1. Bảng 1, 3, 3b: kiểm định chéo lặp (5 lần gập × 10 lần lặp) trên toàn bộ dữ liệu

Thay vì một lần chia 80/20, ta có 50 lần đánh giá. Cột `độ lệch chuẩn` cho biết RMSE dao động bao nhiêu giữa các lần; nếu hai mô hình chênh nhau ít hơn độ lệch chuẩn thì chưa thể kết luận mô hình nào tốt hơn.

```python
def cv_lap_lai(X, y, cot_so, cot_pl, so_lan_lap=10):
    cv = RepeatedKFold(n_splits=5, n_repeats=so_lan_lap, random_state=RANDOM_STATE)
    hang = []
    for ten, mh in tao_mo_hinh().items():
        pipe = tao_pipeline(mh, cot_so, cot_pl)
        kq = cross_validate(pipe, X, y, cv=cv,
                            scoring={"rmse": "neg_root_mean_squared_error", "r2": "r2"})
        hang.append({"Mô hình": ten,
                     "RMSE TB": -kq["test_rmse"].mean(),
                     "Độ lệch chuẩn RMSE": kq["test_rmse"].std(),
                     "R² TB": kq["test_r2"].mean()})
    return pd.DataFrame(hang).set_index("Mô hình")

bai_cat_ngang = {
    "T1":  (t1,  COT_SO_T1,  COT_PL_T1,  MUC_TIEU_T1),
    "T3":  (t3,  COT_SO_T3,  COT_PL_T3,  MUC_TIEU_T3),
    "T3b": (t3b, COT_SO_T3B, COT_PL_T3B, MUC_TIEU_T3B),
}
on_dinh = {}
for k, (d, cot_so, cot_pl, muc_tieu) in bai_cat_ngang.items():
    on_dinh[k] = cv_lap_lai(d[cot_so + cot_pl], d[muc_tieu], cot_so, cot_pl)
    print(f"=== {TEN_BAI[k]}: CV lặp ===")
    display(on_dinh[k].sort_values("RMSE TB").round(3))
```

```python
fig, ax = plt.subplots(1, 3, figsize=(16, 4))
for a, k in zip(ax, on_dinh):
    d = on_dinh[k].sort_values("RMSE TB")
    a.errorbar(d["RMSE TB"], range(len(d)), xerr=d["Độ lệch chuẩn RMSE"], fmt="o", capsize=4)
    a.set_yticks(range(len(d))); a.set_yticklabels(d.index); a.invert_yaxis()
    a.set_xlabel("RMSE trung bình ± độ lệch chuẩn"); a.set_title(TEN_BAI[k], fontsize=10)
plt.tight_layout()
plt.savefig(FIG / "hinh_5_3_on_dinh_cat_ngang.png", dpi=150, bbox_inches="tight")
plt.show()
```

### 5.5.2. Bảng 2: đánh giá nhiều mốc thời gian (rolling origin)

Thay vì một mốc chia duy nhất (2016), ta thử bốn mốc: train đến 2008, 2012, 2016, 2020, mỗi lần dự báo 4 năm kế tiếp. Mô hình **naive** (giữ nguyên giá trị năm cuối của mỗi vùng) làm mốc tham chiếu.

```python
def rolling_origin(moc_cat=(2008, 2012, 2016, 2020), so_nam_test=4):
    hang = []
    for moc in moc_cat:
        tr = t2[t2["Year"] <= moc]
        te = t2[(t2["Year"] > moc) & (t2["Year"] <= moc + so_nam_test)]
        X_tr, y_tr = tr[COT_SO_T2 + COT_PL_T2], tr[MUC_TIEU_T2]
        X_te, y_te = te[COT_SO_T2 + COT_PL_T2], te[MUC_TIEU_T2]
        for ten, mh in tao_mo_hinh().items():
            if ten == BASELINE:
                continue
            pipe = tao_pipeline(mh, COT_SO_T2, COT_PL_T2).fit(X_tr, y_tr)
            hang.append({"Mốc": moc, "Mô hình": ten, "RMSE": rmse(y_te, pipe.predict(X_te))})
        gia_tri_cuoi = tr[tr["Year"] == moc].set_index("Region")[MUC_TIEU_T2]
        hang.append({"Mốc": moc, "Mô hình": "Naive (giữ nguyên năm cuối)",
                     "RMSE": rmse(y_te, te["Region"].map(gia_tri_cuoi))})
    return pd.DataFrame(hang)

kq_roll = rolling_origin()
bang_roll = kq_roll.pivot(index="Mô hình", columns="Mốc", values="RMSE")
bang_roll.columns = [f"train ≤ {c}" for c in bang_roll.columns]
bang_roll["RMSE TB"] = bang_roll.mean(axis=1)

naive = bang_roll.loc["Naive (giữ nguyên năm cuối)"]
cot_moc = [c for c in bang_roll.columns if c.startswith("train")]
bang_roll["Số mốc thắng naive"] = (bang_roll[cot_moc] < naive[cot_moc]).sum(axis=1)
bang_roll.sort_values("RMSE TB").round(3)
```

```python
d = bang_roll.sort_values("RMSE TB")
ax = d["RMSE TB"].plot.barh(figsize=(8, 4), color="#4C72B0")
ax.invert_yaxis(); ax.set_xlabel("RMSE trung bình qua 4 mốc thời gian")
ax.set_title("Bảng 2 – độ chính xác trung bình khi dự báo 4 năm tới")
plt.tight_layout()
plt.savefig(FIG / "hinh_5_4_rolling_origin_bang_2.png", dpi=150, bbox_inches="tight")
plt.show()
```

Qua đề tài, nhóm có thêm kinh nghiệm trong việc lựa chọn dữ liệu, xây dựng quy trình tiền xử lý, lựa chọn biến đầu vào, huấn luyện mô hình và đánh giá kết quả thực nghiệm.

## 6.4. Hạn chế của đề tài

Mặc dù đã hoàn thành các mục tiêu nghiên cứu cơ bản, đề tài vẫn còn một số hạn chế cần được xem xét:

**1. Hạn chế về dữ liệu**

Ba bộ dữ liệu có cấu trúc và phạm vi quan sát khác nhau. Một số thuộc tính quan trọng bị thiếu nhiều hoặc không có dữ liệu, làm giảm số lượng quan sát có thể sử dụng trong quá trình xây dựng mô hình.

Đặc biệt, bộ dữ liệu AIDS regional aggregates thiếu nhiều chỉ số HIV/AIDS quan trọng, khiến phạm vi phân tích và dự báo bị thu hẹp.

**2. Hạn chế về số lượng quan sát**

Một số bài toán chỉ có số lượng quan sát tương đối nhỏ. Điều này có thể khiến kết quả đánh giá thay đổi đáng kể khi thay đổi cách chia dữ liệu hoặc phương pháp kiểm định.

**3. Hạn chế về biến đầu vào**

Các biến đầu vào hiện tại chưa thể hiện đầy đủ những yếu tố có khả năng ảnh hưởng đến tình hình HIV/AIDS, chẳng hạn như chính sách y tế, điều kiện kinh tế, khả năng tiếp cận dịch vụ điều trị và các chương trình phòng chống HIV/AIDS.

**4. Hạn chế về mô hình**

Các mô hình hồi quy sử dụng bộ siêu tham số được thiết lập cố định. Nhóm chưa thực hiện đầy đủ quá trình tìm kiếm và tối ưu siêu tham số bằng những phương pháp như GridSearchCV hoặc RandomizedSearchCV.

Ngoài ra, các mô hình dạng cây như Random Forest và Gradient Boosting có hạn chế trong việc ngoại suy khi dự báo những giá trị nằm ngoài phạm vi dữ liệu đã được huấn luyện.

**5. Hạn chế về phạm vi đánh giá**

Mặc dù nhóm đã thực hiện kiểm định chéo và đánh giá lặp, một số bài toán vẫn có giới hạn về số lượng quan sát và phạm vi thời gian. Vì vậy, kết quả chưa thể đại diện cho toàn bộ diễn biến HIV/AIDS trên phạm vi toàn cầu.

**6. Hạn chế về ý nghĩa ứng dụng**

Các kết quả thu được chủ yếu mang tính phân tích và dự báo thống kê trong phạm vi dữ liệu nghiên cứu. Những kết quả này chưa đủ cơ sở để đưa ra kết luận về quan hệ nhân quả hoặc thay thế đánh giá của các chuyên gia y tế.
